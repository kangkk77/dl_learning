# Jupyter 入门（Day 01）

Notebook 由一个个**单元格（cell）**组成，分两种：
- **Markdown 单元格**：写笔记、公式，比如这一格。双击可以编辑。
- **Code 单元格**：写 Python，运行后结果直接显示在下面。

**最重要的一个快捷键：`Shift + Enter` —— 运行当前格并跳到下一格。**

现在点下面那一格，按 `Shift + Enter` 试试。

```python
print("Hello, Jupyter!")
1 + 1  # 一格里最后一行的值会自动显示，不用 print
```

## 两种模式

| 模式 | 怎么进入 | 光标/边框 | 用途 |
|---|---|---|---|
| 编辑模式 | 点进格子里 / 按 `Enter` | 能打字 | 写代码 |
| 命令模式 | 按 `Esc` | 不能打字 | 用快捷键操作格子 |

**命令模式下的常用快捷键：**

| 按键 | 作用 |
|---|---|
| `A` / `B` | 在上方 / 下方插入新格 |
| `D` `D`（按两次） | 删除当前格 |
| `M` / `Y` | 把当前格变成 Markdown / Code |
| `Z` | 撤销删除格 |
| `↑` / `↓` | 在格子之间移动 |

其它：`Ctrl + Enter` 运行但不跳格；`Alt/Option + Enter` 运行并在下方插新格。

⚠️ **新手最常踩的坑**：来回改、乱序运行之后，内存里的状态和你眼睛看到的代码对不上。

解决办法：菜单 **Kernel → Restart Kernel and Run All Cells**（VS Code 里是顶部的 *Restart* 再 *Run All*），
从头干净地跑一遍。养成习惯：每次学完一节都这么跑一次，能跑通才算数。

## 画图：直接显示在 notebook 里

```python
import matplotlib.pyplot as plt
import numpy as np

# matplotlib 默认字体没有中文，图里要写中文就加这两行（macOS 自带冬青黑体）
plt.rcParams["font.sans-serif"] = ["Hiragino Sans GB", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False  # 让负号正常显示

t = np.linspace(-5, 5, 200)
plt.plot(t, 1 / (1 + np.exp(-t)), label="sigmoid")
plt.plot(t, np.maximum(0, t), label="ReLU")
plt.legend(); plt.grid(True); plt.title("两个常见的激活函数")
plt.show()
```

## "魔法命令"（以 `%` 或 `!` 开头）

- `%timeit 表达式`：测速度
- `!命令`：执行终端命令，比如 `!ls`
- 装包请用 `%pip install 包名`，它保证装进**当前内核**的环境（dl），比 `!pip` 可靠

```python
m = torch.randn(1000, 1000, device=device)
%timeit (m @ m).sum().item()
```
